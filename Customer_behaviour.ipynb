# getting only the numeric Columns Only

identify the Outliers for each Cols

```python
numeric_cols =customer_data.select_dtypes(exclude='object')
```

```python
numeric_cols.shape
```

```python
for x in numeric_cols.columns:
    q1 = numeric_cols[x].quantile(0.25)
    q3 = numeric_cols[x].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    max_bound = q3 + 1.5 * iqr
    
    # تأكد من المحاذاة الصحيحة هنا (4 مسافات)
    numeric_cols_outliers = numeric_cols[(numeric_cols[x] >= lower_bound) & (numeric_cols[x] <= max_bound)]

```

```python
print("the Numbers of Rows in data are " , numeric_cols_outliers.shape[0])
print("the Numbers of Columns in data are " , numeric_cols_outliers.shape[1])
```

```python
numeric_cols_outliers.head()
```

# merging the object table and numeric table after Checking all we need

```python
new_customer_beahvior=pd.concat([object_Customer_col,numeric_cols_outliers],axis =1)
```

```python

new_customer_beahvior = new_customer_beahvior.rename(columns={'Purchase Amount (USD)': 'Purchase_Amount_USD'})
```

```python
from sqlalchemy import create_engine
import urllib.parse

csv_file_path = "D:\\customer_beah_project\\new_customer_beahvior.csv"
table_name = "Customer_Behavior"

server_name = r"DESKTOP-1I8OGM7\SQLEXPRESS"
database_name = "Customer_beahvior_vs"

connection_string = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    f"SERVER={server_name};"
    f"DATABASE={database_name};"
    "Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

try:
    # 4. قراءة ملف CSV وتحميله في الذاكرة
    print("⏳ جاري قراءة ملف الـ CSV...")
    df = pd.read_csv(csv_file_path)
    
    # 5. نقل البيانات بالكامل إلى SQL Server
    print(f"🚀 جاري نقل البيانات إلى جدول [{table_name}]...")
    # if_exists='replace' معناها لو الجدول موجود هيمسحه ويعمله من جديد. لو عايزه يزود عليه غيرها لـ 'append'
    df.to_sql(table_name, con=engine, if_exists='replace', index=False)
    
    print("✅ تم نقل البيانات بنجاح وبدون أي مشاكل!")

except Exception as e:
    print(f"❌ حصلت مشكلة أثناء النقل: {e}")
```
